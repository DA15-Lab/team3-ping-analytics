```python
from google.colab import drive
import pandas as pd
import os

drive.mount('/content/drive')
```

```python
userwithdraw = pd.read_csv('/content/drive/MyDrive/00_원본 csv/accounts_userwithdraw.csv')
```

```python
# admin/test 사유(테스트/운영 계정 탈퇴)는 분석 대상이 아니므로 제외
userwithdraw = userwithdraw[~userwithdraw['reason'].isin(['admin', 'test'])].copy()

# 제외 후 남은 행 수 확인
userwithdraw.shape
```

```python
# created_at을 문자열에서 datetime 타입으로 변환
userwithdraw['created_at'] = pd.to_datetime(userwithdraw['created_at'])
```

```python
# 가장 오래된 날짜 확인 (비정상 과거 날짜가 있는지 확인)
userwithdraw['created_at'].min()
```

```python
# 가장 최근 날짜 확인 (비정상 미래 날짜가 있는지 확인)
userwithdraw['created_at'].max()
```

```python
SAVE_DIR = '/content/drive/MyDrive/전처리_완료_csv'

userwithdraw.to_csv(os.path.join(SAVE_DIR, 'accounts_userwithdraw_cleaned.csv'), index=False)
```
