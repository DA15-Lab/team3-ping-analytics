```python
import os, sys, subprocess

REPO_URL = "https://github.com/DA15-Lab/team3-ping-analytics.git"
REPO_DIR = "/content/team3-ping-analytics"
user_email = "본인_이메일@example.com"
user_name = "본인_이름"

# clone (이미 있으면 건너뜀)
if not os.path.exists(REPO_DIR):
    os.environ['GIT_TERMINAL_PROMPT'] = '0'
    result = subprocess.run(["git", "clone", REPO_URL], capture_output=True, text=True)
    if result.returncode != 0:
        from google.colab import userdata
        token = userdata.get('GITHUB_TOKEN')
        result = subprocess.run(["git", "clone", f"https://{token}@github.com/DA15-Lab/team3-ping-analytics.git"], capture_output=True, text=True)
    print("clone 성공" if result.returncode == 0 else "clone 실패")
else:
    print("이미 clone 되어 있음")

os.chdir(REPO_DIR)

# 👇 여기에 Git 사용자 정보 설정 추가
!git config user.email "{user_email}"
!git config user.name "{user_name}"

# 패키지 설치 (조용히, 이미 맞으면 알아서 스킵)
!pip install -q -r requirements.txt

# 모듈 경로 추가 (중복 방지)
if REPO_DIR + "/src" not in sys.path:
    sys.path.append(REPO_DIR + "/src")

# 드라이브 마운트 (이미 마운트돼 있으면 건너뜀)
from google.colab import drive
if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive')
```
